# Similarity search

Needs the Docker stack running and an experiment that has been ingested. Kernel: the project `.venv`.

```python
%load_ext autoreload
%autoreload 2
from rag_lab.search.quick import ask

# Experiment names are listed on the app's "Try a query" page.
EXPERIMENT = "my-experiment"
OTHER_EXPERIMENT = "my-other-experiment"
```

```python
result = ask("How does multi-head attention work?", EXPERIMENT, chars=100)
```

Only table chunks. `result.hits[0].text` has the full text of a hit.

```python
result = ask("BLEU scores", EXPERIMENT, top_k=3, modality="table", chars=100)
print(result.hits[0].text)
```

The same query against two experiments. Set `OTHER_EXPERIMENT` to a second experiment name.

```python
for name in [EXPERIMENT, OTHER_EXPERIMENT]:
    print(f"--- {name}")
    ask("What optimizer was used for training?", name, top_k=3)
```
